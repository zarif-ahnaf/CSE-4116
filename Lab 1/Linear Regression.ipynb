## Hypothesis Function
In linear regression, the hypothesis function is the equation used to make predictions about the dependent variable based on the independent variables. It represents the relationship between the input features and the target output.

For a simple case with one independent variable, the hypothesis function is:
```
h(x)=β₀+β₁x
```

Where:

* h(x) is the predicted value of the dependent variable (y).
* x is the independent variable.
* β₀ is the intercept, representing the value of y when x is 0.
* β₁ is the slope, indicating how much y changes for each unit change in x.

For multiple linear regression (with more than one independent variable), the hypothesis function expands to:
```
h(x₁,x₂,...,xₖ)=β₀+β₁x₁+β₂x₂+...+βₖxₖ
```

Where:
* x₁,x₂,...,xₖ are the independent variables.
* β₀ is the intercept.
* β₁,β₂,...,βₖ are the coefficients, representing the influence of each respective independent variable on the predicted output.

## Regression Evaluation Metrics


Here are three common evaluation metrics for regression problems:

**Mean Absolute Error** (MAE) is the mean of the absolute value of the errors:

$$\frac 1n\sum_{i=1}^n|y_i-\hat{y}_i|$$

**Mean Squared Error** (MSE) is the mean of the squared errors:

$$\frac 1n\sum_{i=1}^n(y_i-\hat{y}_i)^2$$

**Root Mean Squared Error** (RMSE) is the square root of the mean of the squared errors:

$$\sqrt{\frac 1n\sum_{i=1}^n(y_i-\hat{y}_i)^2}$$

**R-squared** (R²) is the proportion of the variance in the dependent variable that is predictable from the independent variables. It indicates how well the model fits the observed data. A higher R-squared value indicates a better fit. The formula is:

$$R^2 = 1 - \frac{\sum_{i=1}^n (y_i - \hat{y}_i)^2}{\sum_{i=1}^n (y_i - \bar{y})^2}$$

Comparing these metrics:

- **MAE** is the easiest to understand, because it's the average error.
- **MSE** is more popular than MAE, because MSE "punishes" larger errors, which tends to be useful in the real world.
- **RMSE** is even more popular than MSE, because RMSE is interpretable in the "y" units.
- **R-squared** provides an indication of the goodness of fit and is a useful measure for comparing models.


All of these are **loss functions**, because you want to minimize them.

## 4. Dividing into Feature (X) and Target (y)

````python
X = df[['Avg. Area Income', 'Avg. Area House Age', 'Avg. Area Number of Rooms',
               'Avg. Area Number of Bedrooms', 'Area Population']]
y = df['Price']
````

## 5.1 Train - Test Split
Now let's split the data into a training set and a testing set. You will train out model on the training set and then use the test set to evaluate the model. Note that you are using 40% of the data for testing.

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state=random_state)
````

## 5.2 Normalization

Normalization (often called standardization) is a crucial preprocessing step in machine learning, especially for models that are sensitive to the scale of input features, like Linear Regression. It involves transforming numerical features to a common scale without distorting differences in the ranges of values or losing information.

### Why Normalize?

1.  **Prevent Feature Dominance**: Features with larger values might disproportionately influence the model's objective function, making the model biased towards them. Normalization ensures that all features contribute equally to the model.
2.  **Improve Model Performance**: Many machine learning algorithms, particularly those that use gradient descent (like Linear Regression), converge faster and perform better when features are on a similar scale.
3.  **Handle Outliers**: While not a primary goal, some normalization techniques can make the model more robust to outliers.

### StandardScaler

`StandardScaler` from Scikit-Learn is a common normalization technique. It transforms the data such that its distribution has a mean of 0 and a standard deviation of 1. The formula for standardization for a data point `x` is:

```
z = (x - μ) / σ
```

Where:
*   `μ` (mu) is the mean of the feature.
*   `σ` (sigma) is the standard deviation of the feature.

This method is particularly effective when the data follows a Gaussian (normal) distribution. After applying `StandardScaler`, the `X_train` and `X_test` datasets will have their features scaled, ensuring that the linear regression model is trained on a more consistent data representation.

````python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

## 6. Creating and Training the Model

````python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
````

````python
print("Intercept:", model.intercept_)
````

````python
coeff_df = pd.DataFrame(model.coef_,X.columns,columns=['Coefficient'])
coeff_df
````

````python
print('Interpreting the coefficients:')
for index, row in coeff_df.iterrows():
    feature = index
    coefficient = row['Coefficient']

    # Determine if it's an increase or decrease
    change_type = 'increase' if coefficient >= 0 else 'decrease'

    # Format the coefficient value for display
    formatted_coefficient = f'${abs(coefficient):,.2f}'

    print(f"- Holding all other features fixed, a 1 unit increase in {feature} is associated with an {change_type} of {formatted_coefficient}.")
````

## 7. Model Evaluation

````python
predictions = model.predict(X_test)
plt.figure(figsize=(10, 6))
plt.scatter(y_test, predictions, s=20, facecolors='none', edgecolors='blue')
plt.xlabel("True Prices (y_test)")
plt.ylabel("Predicted Prices (predictions)")
plt.title("True Prices vs. Predicted Prices")
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', label='Zero Error Line')
plt.legend()
plt.show()
````

````python
from sklearn import metrics
print('MAE      :', metrics.mean_absolute_error(y_test, predictions))
print('MSE      :', metrics.mean_squared_error(y_test, predictions))
print('RMSE     :', np.sqrt(metrics.mean_squared_error(y_test, predictions)))
print('R² Score :', np.sqrt(metrics.r2_score(y_test, predictions)))
````
