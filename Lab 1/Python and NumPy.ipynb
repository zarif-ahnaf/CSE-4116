## 1.3 Loops — Processing Every Data Point

```python
# Basic loop with range
for i in range(len(names)):
    print(f"{names[i]} studied {hours[i]}h and scored {scores[i]}")
```

Filtering is a core preprocessing step in every ML pipeline.

```python
# Keep only students who studied more than 5 hours
high_effort = [name for name, h in zip(names, hours) if h > 5]
print("High effort students:", high_effort)   # ['Bob', 'David', 'Eva']
```

## 1.5 Functions — Reusable ML Steps

```python
def passing_rate(scores, threshold=70):
    """Returns the fraction of students who passed."""
    passed = [s for s in scores if s >= threshold]
    return len(passed) / len(scores)

rate = passing_rate(scores)
print(f"Passing rate (>=70): {rate:.0%}")   # 80%

# Default argument in action
print(f"Distinction rate (>=85): {passing_rate(scores, threshold=85):.0%}")
```

```python
# Experiment 1.5
# Write a function mean_score(scores) that returns the average score.
# Then write range_score(scores) that returns max - min.

def mean_score(scores):
    return sum(scores) / len(scores)

def range_score(scores):
    return max(scores) - min(scores)

print("Mean:", mean_score(scores))
print("Range:", range_score(scores))
```

# Part 2 — NumPy
## 2.1 Creating Arrays
NumPy arrays are the backbone of every ML library. Let's rebuild our dataset as arrays.

```python
import numpy as np

hours_arr  = np.array([5, 8, 3, 10, 6])
scores_arr = np.array([72, 85, 60, 92, 78])

print(hours_arr.shape)   # (5,)  → 1D, 5 elements
print(hours_arr.dtype)   # int64
```

```python
# 2D feature matrix: rows = students, columns = features
X = np.array([
    [5,  72],
    [8,  85],
    [3,  60],
    [10, 92],
    [6,  78]
])

print("Shape:",  X.shape)      # (5, 2)
print("Alice:",  X[0])         # [5, 72]
print("All hours:",  X[:, 0])  # column 0
print("All scores:", X[:, 1])  # column 1
```

```python
# Experiment 2.1
# Create a 3rd column for sleep hours and add it to X.
# What does np.zeros((5,3)) look like? And np.ones((5,3))?

sleep = np.array([7, 6, 5, 8, 7])
X3 = np.column_stack([X, sleep])
print("X with sleep:\n", X3)
print("New shape:", X3.shape)

print("\nnp.zeros(5,3):\n", np.zeros((5, 3)))
```

---
## 2.2 Array Operations — No Loops Needed

```python
# Min-max normalisation (a taste of feature scaling)
normalized = (hours_arr - hours_arr.min()) / (hours_arr.max() - hours_arr.min())
print("Normalised hours:", normalized.round(2))   # [0.29 0.71 0.   1.   0.43]

# Bonus 5 points to everyone
boosted = scores_arr + 5
print("Boosted scores:", boosted)

# Element-wise multiply (weighted sum preview)
weights = np.array([0.4, 0.3, 0.2, 0.1, 0.0])
print("Weighted scores:", scores_arr * weights)
```

```python
# Experiment 2.2
# Apply z-score normalisation to scores_arr:
#   z = (x - mean) / std
# The result should have mean ≈ 0 and std ≈ 1. Verify!

z = (scores_arr - scores_arr.mean()) / scores_arr.std()
print("Z-scores:",   z.round(2))
print("Mean of z:",  z.mean().round(6))   # should be ~0
print("Std of z:",   z.std().round(6))    # should be ~1
```

## 2.3 Statistical Functions — Describing Your Data

```python
print("Mean:  ", scores_arr.mean())        # 77.4
print("Std:   ", scores_arr.std().round(2))# 11.07
print("Min:   ", scores_arr.min())         # 60
print("Max:   ", scores_arr.max())         # 92
print("Median:", np.median(scores_arr))    # 78.0

# Per-column stats on the 2D matrix
print("\nRow means:", X.mean(axis=1))      # [38.5, 46.5, 31.5, 51. , 42. ]
print("\nColumn means:", X.mean(axis=0))   # [6.4  77.4]
```

## 2.7 Matrix Operations — The Core of ML Math

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(A)
print(B)
print("Element-wise (A * B):\n", A * B)
print("\nMatrix multiply (A @ B):\n", A @ B)
print("\nTranspose A.T:\n", A.T)
```

```python
print(A)
print(B)
```

```python
# Manual linear prediction: score = 50 + 4 × hours
w = np.array([50, 4])   # [intercept, slope]

X_bias = np.column_stack([np.ones(5), hours_arr])   # add bias column of 1s
print("Feature matrix with bias:\n", X_bias)

predictions = X_bias @ w
print("\nPredictions:", predictions)
print("Actual:     ", scores_arr)
print("Errors:     ", scores_arr - predictions)
```

```python
# Experiment 2.7
# Tweak the intercept and slope to minimise the errors.
# Can you get the total absolute error below 15?

intercept = 50   # ← change me
slope     = 4    # ← change me

w_new = np.array([intercept, slope])
preds = X_bias @ w_new
errors = scores_arr - preds
print("Errors:     ", errors)
print("Total abs error:", np.abs(errors).sum())
```

## 2.8 Train / Test Split — The NumPy Way

```python
np.random.seed(42)

n       = len(scores_arr)
indices = np.arange(n)
np.random.shuffle(indices)

split    = int(0.8 * n)          # 80% train
train_idx = indices[:split]
test_idx  = indices[split:]

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = scores_arr[train_idx], scores_arr[test_idx]

print("Train indices:", train_idx)
print("Test  indices:", test_idx)
print("X_train shape:", X_train.shape)
print("X_test shape: ", X_test.shape)
```

```python
# Experiment 2.8
# Change the seed to 0, 1, 7 — observe how the split changes.
# Change the split ratio to 0.6 (60/40 split). How big is X_test now?

np.random.seed(0)    # ← try different seeds
split_ratio = 0.6    # ← try 0.6 or 0.7

idx = np.arange(n)
np.random.shuffle(idx)
sp = int(split_ratio * n)
print("Train:", idx[:sp], "| Test:", idx[sp:])
```
